query = f"SELECT {selected_columns} FROM {table}"

    conditions = []

```python
import math

# ---------------- Assignment 4 ----------------

def student_report(name, *marks, **kwargs):
    total = sum(marks)
    average = total / len(marks)
    passed = all(mark >= 35 for mark in marks)

    print("\n------ STUDENT REPORT ------")
    print("Student :", name)
    print("Course  :", kwargs.get("course", "Not provided"))
    print("City    :", kwargs.get("city", "Not provided"))
    print("Total   :", total)
    print("Average :", average)

    if passed:
        print("Result  : PASS")
    else:
        print("Result  : FAIL")


# ---------------- Assignment 9 ----------------

def checkout(*prices, **kwargs):
    subtotal = sum(prices)

    discount = kwargs.get("discount", 0)
    tax = kwargs.get("tax", 0)

    discount_amount = subtotal * discount / 100
    amount_after_discount = subtotal - discount_amount

    tax_amount = amount_after_discount * tax / 100
    final_amount = amount_after_discount + tax_amount

    print("\n------ SHOPPING BILL ------")
    print("Customer :", kwargs.get("customer", "Guest"))
    print("Subtotal :", round(subtotal, 2))
    print("Discount :", round(discount_amount, 2))
    print("Tax      :", round(tax_amount, 2))
    print("Final Bill :", round(final_amount, 2))


# ---------------- Assignment 17 ----------------

def etl_summary(source, destination, *tables, **kwargs):

    row_counts = kwargs.get("row_counts", {})
    batch_size = kwargs.get("batch_size", 1000)

    table_count = len(tables)

    total_rows = sum(
        row_counts.get(table, 0)
        for table in tables
    )

    batches = math.ceil(total_rows / batch_size) if total_rows else 0

    print("\n------ ETL SUMMARY ------")
    print("Source :", source)
    print("Destination :", destination)
    print("Tables :", tables)
    print("Table Count :", table_count)
    print("Total Rows :", total_rows)
    print("Required Batches :", batches)


# ---------------- Assignment 20 ----------------

def process_order(order_id, *products, **kwargs):

    subtotal = 0
    total_quantity = 0

    for product in products:

        if not isinstance(product, dict):
            continue

        price = product.get("price", 0)
        quantity = product.get("quantity", 0)

        subtotal += price * quantity
        total_quantity += quantity

    discount = kwargs.get("discount", 0)
    tax = kwargs.get("tax", 0)

    discount_amount = subtotal * discount / 100
    amount_after_discount = subtotal - discount_amount

    tax_amount = amount_after_discount * tax / 100
    final_amount = amount_after_discount + tax_amount

    print("\n------ ORDER SUMMARY ------")
    print("Order ID :", order_id)
    print("Customer :", kwargs.get("customer", "Guest"))
    print("Payment Mode :", kwargs.get("payment_mode", "Not Provided"))
    print("Total Quantity :", total_quantity)
    print("Subtotal :", subtotal)
    print("Discount :", discount_amount)
    print("Tax :", tax_amount)
    print("Final Amount :", final_amount)


# ================= MENU =================

while True:

    print("\n========== MAIN MENU ==========")
    print("1. Student Report")
    print("2. Shopping Cart")
    print("3. ETL Summary")
    print("4. Order Processing")
    print("5. Exit")

    choice = input("Enter your choice: ")

    # ---------------- Option 1 ----------------

    if choice == "1":

        name = input("Enter Student Name: ")
        course = input("Enter Course: ")
        city = input("Enter City: ")

        count = int(input("How many subjects? "))

        marks = []

        for i in range(count):
            mark = int(input(f"Enter Mark {i+1}: "))
            marks.append(mark)

        student_report(name, *marks,
                       course=course,
                       city=city)

    # ---------------- Option 2 ----------------

    elif choice == "2":

        customer = input("Customer Name: ")

        n = int(input("Number of Products: "))

        prices = []

        for i in range(n):
            price = float(input(f"Price {i+1}: "))
            prices.append(price)

        discount = float(input("Discount %: "))
        tax = float(input("Tax %: "))

        checkout(*prices,
                 customer=customer,
                 discount=discount,
                 tax=tax)

    # ---------------- Option 3 ----------------

    elif choice == "3":

        source = input("Source: ")
        destination = input("Destination: ")

        table_count = int(input("Number of Tables: "))

        tables = []
        row_counts = {}

        for i in range(table_count):

            table = input(f"Table {i+1} Name: ")
            rows = int(input("Row Count: "))

            tables.append(table)
            row_counts[table] = rows

        batch_size = int(input("Batch Size: "))

        etl_summary(
            source,
            destination,
            *tables,
            row_counts=row_counts,
            batch_size=batch_size
        )

    # ---------------- Option 4 ----------------

    elif choice == "4":

        order_id = input("Order ID: ")
        customer = input("Customer Name: ")
        payment = input("Payment Mode: ")

        n = int(input("Number of Products: "))

        products = []

        for i in range(n):

            name = input("Product Name: ")
            price = float(input("Price: "))
            quantity = int(input("Quantity: "))

            products.append({
                "name": name,
                "price": price,
                "quantity": quantity
            })

        discount = float(input("Discount %: "))
        tax = float(input("Tax %: "))

        process_order(
            order_id,
            *products,
            customer=customer,
            payment_mode=payment,
            discount=discount,
            tax=tax
        )

    # ---------------- Exit ----------------

    elif choice == "5":
        print("Program Closed Successfully.")
        break

    else:
        print("Invalid Choice! Try Again.")
```
