Accept a non-negative integer.

Calculate its factorial using a loop.

Return the factorial.

Handle negative input.

```python
def calculate_factorial(n):
    # Handle negative input
    if n < 0:
        return "Factorial is not defined for negative numbers"

    factorial = 1

    # Calculate factorial using loop
    for i in range(1, n + 1):
        factorial *= i

    return factorial


# Accept input from user
num = int(input("Enter a non-negative integer: "))

# Call the function and print result
result = calculate_factorial(num)
print("Factorial:", result)
```

```python
def reverse_string_slicing(text):
    return text[::-1]

def reverse_string_loop(text):
    reversed_text = ""

    for char in text:
        reversed_text = char + reversed_text

    return reversed_text

text = input("Enter text: ")

print("Reversed using slicing:", reverse_string_slicing(text))
print("Reversed using loop:", reverse_string_loop(text))
```

# Assignment 19: Find Maximum in a List
Write a function named find_list_maximum().

Requirements:

Accept a list of numbers.

Do not use max().

Return the largest list element.

Handle an empty list by returning an appropriate message.

```python
# Assignment 19: Find Maximum in a List

def find_list_maximum(numbers):
    if len(numbers) == 0:
        return None

    maximum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum

try:
    numbers = list(map(float, input("Enter numbers separated by space: ").split()))

    result = find_list_maximum(numbers)

    if result is None:
        print("The list is empty.")
    else:
        print("Maximum value:", result)

except ValueError:
    print("Please enter valid numbers.")
```

Write a function named count_number_types().

Return all three counts.

Use tuple unpacking to receive the returned values.

```python
# Assignment 20: Count Positive, Negative and Zero Values

def count_number_types(numbers):
    positive = 0
    negative = 0
    zeros = 0

    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zeros += 1

    return positive, negative, zeros

try:
    numbers = list(map(int, input("Enter integers separated by space: ").split()))

    positive, negative, zeros = count_number_types(numbers)

    print("Positive numbers:", positive)
    print("Negative numbers:", negative)
    print("Zeros:", zeros)

except ValueError:
    print("Please enter valid integers.")
```

# Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements:

Accept a string.

Return a dictionary containing:

Total characters

Total words

Total vowels

Total digits

Total spaces

Display every result clearly.

```python
# Function to analyze text
def analyze_text(text):
    vowels = "aeiouAEIOU"

    result = {
        "Total characters": len(text),
        "Total words": len(text.split()),
        "Total vowels": sum(1 for ch in text if ch in vowels),
        "Total digits": sum(1 for ch in text if ch.isdigit()),
        "Total spaces": text.count(" ")
    }

    return result

# User input
text = input("Enter text: ")

# Get analysis
analysis = analyze_text(text)

# Display results
for key, value in analysis.items():
    print(f"{key}: {value}")
```

# Assignment 34: Prime Number Function
Write a function named is_prime().

Requirements:

Accept one integer.

Return True when the number is prime.

Return False otherwise.

Use the function to print all prime numbers from 1 to 100.

```python
# Function to check prime number
def is_prime(number):
    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True

# Print prime numbers from 1 to 100
print("Prime numbers from 1 to 100:")

for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

# Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements:

Accept the number of terms.

Return the complete Fibonacci series as a list.

Handle zero and negative term counts.

Do not use recursion in this assignment.

```python
# Function to generate Fibonacci series
def generate_fibonacci(n):
    if n < 0:
        return "Number of terms cannot be negative."

    if n == 0:
        return []

    if n == 1:
        return [0]

    series = [0, 1]

    for i in range(2, n):
        series.append(series[-1] + series[-2])

    return series

# User input
terms = int(input("Enter number of terms: "))

# Display result
result = generate_fibonacci(terms)
print("Fibonacci series:", result)
```

# Assignment 92: Closure-Based Bank Balance

Use nonlocal to update the balance.

Reject withdrawals greater than the available balance.

Return all three inner functions.

```python
def create_bank_account(initial_balance=0):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount")

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def check_balance():
        print("Current Balance:", balance)

    return deposit, withdraw, check_balance

# Create account
deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()
deposit(500)
check_balance()
withdraw(300)
check_balance()
withdraw(2000)
check_balance()
```

# Assignment 93: Function Decorator
Create a decorator named execution_logger().

Requirements:

The decorator should print a message before the original function executes.

It should print another message after execution.

Apply it to an addition function.

Preserve the original function's return value.

```python
# Decorator function
def execution_logger(func):

    def wrapper(*args, **kwargs):
        print("Starting function...")

        result = func(*args, **kwargs)

        print("Function completed.")

        return result

    return wrapper

# Apply decorator
@execution_logger
def add(a, b):
    result = a + b
    print("Result:", result)
    return result

# Call function
final_result = add(10, 20)

print("Returned Value:", final_result)
```

```python
from functools import reduce

# -----------------------------
# Global variable (LEGB Demo)
# -----------------------------
company_name = "Edukron"

# -----------------------------
# Employee Data
# -----------------------------
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]

# -----------------------------
# Part 1: Display Employees
# -----------------------------
def display_employees(employee_list):
    if not employee_list:
        print("No employees found.")
        return

    for emp in employee_list:
        print("-" * 40)
        print(f"ID: {emp['id']}")
        print(f"Name: {emp['name']}")
        print(f"Department: {emp['department']}")
        print(f"Salary: {emp['salary']}")
        print(f"Performance: {emp['performance_score']}")
        print(f"Experience: {emp['experience']} years")
    print("-" * 40)

# -----------------------------
# Part 2: Extract Employee Names
# -----------------------------
def get_employee_names():
    return list(map(lambda e: e["name"], employees))

# -----------------------------
# Part 3: Filter High Performers
# -----------------------------
def get_high_performers():
    return list(filter(lambda e: e["performance_score"] >= 80, employees))

# -----------------------------
# Part 4: Filter by Department
# -----------------------------
def search_by_department(department):
    result = list(
        filter(lambda e: e["department"].lower() == department.lower(), employees)
    )
    return result

# -----------------------------
# Part 5: Salary Increase
# -----------------------------
def calculate_revised_salaries():
    def revise(emp):
        score = emp["performance_score"]

        if score >= 90:
            increase = 15
        elif score >= 80:
            increase = 10
        elif score >= 70:
            increase = 5
        else:
            increase = 0

        new_salary = emp["salary"] + (emp["salary"] * increase / 100)

        return {
            "name": emp["name"],
            "old_salary": emp["salary"],
            "increase_percent": increase,
            "new_salary": round(new_salary, 2)
        }

    return list(map(revise, employees))

# -----------------------------
# Part 6: Total Salary Expense
# -----------------------------
def salary_expense_report():
    current_total = reduce(
        lambda acc, e: acc + e["salary"],
        employees,
        0
    )

    revised = calculate_revised_salaries()

    revised_total = reduce(
        lambda acc, e: acc + e["new_salary"],
        revised,
        0
    )

    additional = revised_total - current_total

    return current_total, revised_total, round(additional, 2)

# -----------------------------
# Part 7: Highest-Paid Employee
# -----------------------------
def highest_paid_employee():
    return reduce(
        lambda a, b: a if a["salary"] >= b["salary"] else b,
        employees
    )

# -----------------------------
# Part 8: Sort Employees
# -----------------------------
def sort_employees(option):
    if option == "name":
        return sorted(employees, key=lambda e: e["name"])

    elif option == "salary":
        return sorted(employees, key=lambda e: e["salary"])

    elif option == "performance":
        return sorted(employees, key=lambda e: e["performance_score"], reverse=True)

    elif option == "experience":
        return sorted(employees, key=lambda e: e["experience"], reverse=True)

    else:
        return employees

# -----------------------------
# Part 9: Closure-Based Bonus Calculator
# -----------------------------
def create_bonus_calculator(percent):

    def calculate_bonus(salary):
        return round(salary * percent / 100, 2)

    return calculate_bonus

bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)

# -----------------------------
# Part 10 & 11: Nested Report Generator + LEGB
# -----------------------------
def generate_department_report(department_name):

    # Enclosing scope variable
    department = department_name

    dept_employees = search_by_department(department)

    if not dept_employees:
        return {"error": "Department not found"}

    # Local variable
    report_title = f"{department} Department Report"

    def employee_count():
        return len(dept_employees)

    def total_salary():
        return sum(map(lambda e: e["salary"], dept_employees))

    def average_salary():
        return round(total_salary() / employee_count(), 2)

    def average_performance():
        total = sum(map(lambda e: e["performance_score"], dept_employees))
        return round(total / employee_count(), 2)

    report = {
        "company": company_name,      # Global scope
        "department": department,     # Enclosing scope
        "title": report_title,        # Local scope
        "employee_count": employee_count(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_performance": average_performance()
    }

    return report

# -----------------------------
# Test Calls (Optional)
# -----------------------------
if __name__ == "__main__":

    print("\nEmployee Names:")
    print(get_employee_names())

    print("\nHigh Performers:")
    display_employees(get_high_performers())

    print("\nIT Department:")
    display_employees(search_by_department("IT"))

    print("\nRevised Salaries:")
    for item in calculate_revised_salaries():
        print(item)

    print("\nSalary Expense Report:")
    current, revised, extra = salary_expense_report()
    print("Current:", current)
    print("Revised:", revised)
    print("Additional:", extra)

    print("\nHighest Paid Employee:")
    print(highest_paid_employee())

    print("\nSorted by Salary:")
    display_employees(sort_employees("salary"))

    print("\nBonus on 50000:")
    print("5%:", bonus_5(50000))
    print("10%:", bonus_10(50000))
    print("15%:", bonus_15(50000))

    print("\nDepartment Report:")
    print(generate_department_report("IT"))
```
