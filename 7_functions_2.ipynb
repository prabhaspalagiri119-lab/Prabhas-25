## **Section 5: Combined Lambda, map(), filter(), and reduce() — 20 Questions**

```python
## Question 1
## Using filter() and map():
## Extract all even numbers.
## Calculate the square of every even number.
## Return the squared values as a list.

from functools import reduce

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

result = list(
    map(
        lambda x: x ** 2,
        filter(lambda x: x % 2 == 0, numbers)
    )
)

print(result)
```

```python
## Question 2
## Using filter() and reduce():
## Extract all odd numbers.
## Calculate their total sum.


from functools import reduce

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

result = reduce(
    lambda x, y: x + y,
    filter(lambda x: x % 2 != 0, numbers)
)

print(result)
```

```python
## Question 3
## Using filter() and map():
## Extract numbers greater than 50.
## Multiply every selected number by 2.

numbers = [10, 25, 40, 55, 70, 85, 100]

result = list(
    map(
        lambda x: x * 2,
        filter(lambda x: x > 50, numbers)
    )
)

print(result)
```

```python
## Question 4
## Using filter(), map(), and reduce():
## Extract positive numbers.
## Calculate the square of every positive number.
## Calculate the total sum of the squared values.

from functools import reduce

numbers = [-15, 20, -35, 40, 50, -10, 60]

result = reduce(
    lambda x, y: x + y,
    map(
        lambda x: x ** 2,
        filter(lambda x: x > 0, numbers)
    )
)

print(result)
```

```python
## Question 14
## Using filter() and reduce():
## Extract students who passed.
## Find the topper among the passed students.

from functools import reduce

numbers = [5, 10, 15, 20, 25]

result = reduce(
    lambda x, y: x + y,
    filter(lambda x: x > 10, numbers)
)

print(result)
```

```python
## Question 15
## Using filter(), map(), and reduce():
## Extract students scoring more than 70.
## Extract only their marks.
## Calculate the average marks.

numbers = [2, 5, 8, 11, 14, 17, 20]

result = list(
    map(
        lambda x: x ** 2,
        filter(lambda x: x > 10, numbers)
    )
)

print(result)
```

```python
## Question 16
## Using filter(), map(), and reduce():
## Extract products currently in stock.
## Calculate each product's inventory value using price × stock.
## Calculate the total inventory value.

numbers = [10, 15, 20, 25, 30, 35, 40]

result = list(
    map(
        lambda x: x - 5,
        filter(lambda x: x % 10 == 0, numbers)
    )
)

print(result)
```

```python
## Question 17
## Using filter(), map(), and reduce():
## Extract products belonging to the Electronics category.
## Apply a 20% discount.
## Calculate the total discounted value.

from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(
    lambda x, y: x * y,
    filter(lambda x: x % 2 == 1, numbers)
)

print(result)
```

```python
## Question 18
## Using filter() and reduce():
## Extract delivered orders.
## Calculate the total delivered revenue.

names = ["python", "java", "c", "javascript", "go"]

result = list(
    map(
        lambda x: x.upper(),
        filter(lambda x: len(x) >= 4, names)
    )
)

print(result)
```

```python
## Question 19
## Using filter(), map(), and reduce():
## Extract transactions whose amount is greater than 5000.
## Add 18% GST to every selected transaction.
## Calculate the grand total after GST.

from functools import reduce

numbers = [8, 12, 16, 20]

result = reduce(
    lambda x, y: x + y,
    map(lambda x: x // 4, numbers)
)

print(result)
```
