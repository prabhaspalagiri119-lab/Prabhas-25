## 1. Simple Line Plot

Create a simple line plot using X and Y values.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y, color='red', marker='o', linewidth=2)

plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.show()
```

### Analysis:
- The graph represents a continuous increase in the values.
- The first observation has the smallest Y value.
- The final observation records the maximum Y value.
- The change between the last two points is comparatively larger.
- Overall, the data indicates positive growth.

### 2. Seaborn Line Plot
Create a line plot showing the number of passengers over time

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.figure(figsize=(6, 4))

plt.plot(
    x,
    y,
    linestyle='--',
    color='blue',
    label='Sales'
)

plt.legend()
plt.tight_layout()

plt.show()
```

### Analysis:
- The dashed line makes the movement of the values easy to identify.
- Sales values progress upward across the given observations.
- There are no sudden downward movements in the series.
- The final point indicates the strongest sales value.
- The pattern suggests consistent improvement in sales

## 3. Seaborn Box Plot

Create a box plot to compare bill amounts across different days.


```python
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset('flights')

sns.lineplot(
    data=flights,
    x='year',
    y='passengers'
)

plt.title('Passengers over Time')

plt.show()
```

### Analysis:
- Passenger traffic changes significantly across the years.
- The dataset shows repeated rises and falls within the time period.
- The later portion of the graph reaches noticeably higher levels.
- This pattern indicates that passenger demand has expanded over time.
- The recurring fluctuations suggest that time-based factors influence travel demand.

### 4. Plotly Iris Scatter Plot

Create a scatter plot using the Iris dataset and differentiate species using colors.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.boxplot(
    data=tips,
    x='day',
    y='total_bill',
    palette='Set1'
)

plt.title('Bill Amount by Day')

plt.show()
```

### Analysis:
- The distribution of bills is not identical for all days.
- Different median positions indicate differences in typical spending.
- Some days have a wider spread of bill values than others.
- A few observations extend beyond the normal range of the distributions.
- Customer spending therefore varies depending on the day.

### 5. Plotly Line Chart – India Life Expectancy

Create a line chart showing life expectancy in India over the years.

```python
import plotly.express as px

df = px.data.iris()

fig = px.scatter(
    df,
    x='sepal_width',
    y='sepal_length',
    color='species',
    color_discrete_map={
        'setosa': 'red',
        'versicolor': 'blue',
        'virginica': 'green'
    }
)

fig.show()
```

### Analysis:
- The observations form separate groups based on flower species.
- Setosa appears more compact compared with the other groups.
- Some observations from different species occupy nearby regions.
- The plotted features therefore provide useful information for species comparison.
- However, complete classification cannot be achieved using these two measurements alone.

### 6. Life Expectancy in India

```python
import plotly.express as px

df = px.data.gapminder().query("country == 'India'")

fig = px.line(
    df,
    x='year',
    y='lifeExp',
    title='Life Expectancy in India'
)

fig.update_traces(
    line=dict(color='violet', width=3)
)

fig.show()
```

### Analysis:
- Life expectancy shows a long-term upward movement in the graph.
- The values generally become higher as the years progress.
- There are no major long-term declines visible in the series.
- The increasing values indicate improvement over the observed period.
- Overall, the chart reflects a positive development in life expectancy.

## 7.Pie Chart

Create a Pie Chart to show the
distribution of students in different departments.

```python
import matplotlib.pyplot as plt

departments = ['Computer Science', 'Data Science', 'Commerce', 'Management']
students = [40, 30, 20, 10]

plt.pie(
    students,
    labels=departments,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Students Distribution by Department')

plt.show()
```

### Analysis:
- Computer Science has the largest share of students.
- Data Science represents the second-largest group.
- Commerce contributes a smaller portion of the total.
- Management has the lowest number of students.
- The pie chart clearly shows the proportion of students across departments.
