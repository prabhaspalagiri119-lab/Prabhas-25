## Arithmetic Operator

```python
# Write a program to perform addition, subtraction, multiplication, division, floor division, modulus, and exponentiation on two numbers.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Modulus =", a % b)
print("Exponent =", a ** b)
```

```python
# Swap two numbers without using a third variable.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

a, b = b, a

print("a =", a)
print("b =", b)
```

```python
#  Find whether a number is even or odd using the modulus operator.

num = int(input("Enter number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Reverse a three-digit number using arithmetic operators.

num = int(input("Enter number: "))

reverse = (num % 10) * 100 + ((num // 10) % 10) * 10 + (num // 100)

print(reverse)
```

```python
# Find the last digit and first digit of a given number.

num = int(input("Enter number: "))

last = num % 10

while num >= 10:
    num //= 10

print("First Digit =", num)
print("Last Digit =", last)
```

## **Assignment Operators**

```python
#  Demonstrate all assignment operators (=, +=, -=, *=, /=, //=, %= and **=) on a variable.

x = 10

print(x)

x += 5
print(x)

x -= 2
print(x)

x *= 3
print(x)

x /= 2
print(x)

x //= 2
print(x)

x %= 4
print(x)

x **= 2
print(x)
```

```python
# Update a bank account balance using compound assignment operators.

balance = 1000

deposit = int(input("Deposit: "))
balance += deposit

withdraw = int(input("Withdraw: "))
balance -= withdraw

print("Balance =", balance)
```

```python
#  Calculate the total marks using += inside a loop.

total = 0

for i in range(5):
    mark = int(input("Enter marks: "))
    total += mark

print(total)
```

# **Logical Operators**

```python
# Check if a person is eligible for a loan (age > 21 AND salary > 30000).

age = int(input("Enter your age:"))
salary = int(input("Enter your Salary:"))

if age >= 25:
  if salary >= 30000:
    print("Eligible for loan")
else:
  print("Not Eligible to loan")
```

```python
# Determine whether a year is a leap year using logical conditions.

year = int(input("Enter year:"))
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
  print("Leap year")
else:
  print("Not Leap year")
```

## **Bitwise Operators**

```python
 # Find the bitwise AND, OR, XOR of two numbers.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("AND =", a & b)
print("OR =", a | b)
print("XOR =", a ^ b)
```

```python
# Perform left shift and right shift operations.

num = int(input("Enter number: "))

print("Left Shift =", num << 1)
print("Right Shift =", num >> 1)
```

```python
# Check whether a number is even using a bitwise operator.

num = int(input("Enter number: "))

if (num & 1) == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Swap two numbers using the XOR operator.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

a = a ^ b
b = a ^ b
a = a ^ b

print("a =", a)
print("b =", b)
```

```python
# Count how many bits are set to 1 in a number using bitwise operations.

num = int(input("Enter number: "))

count = 0

while num:
    count += num & 1
    num >>= 1

print(count)
```

```python
# Count how many vowels in a string belong to a predefined vowel list using in.

text = input("Enter string: ")

vowels = "aeiouAEIOU"
count = 0

for ch in text:
    if ch in vowels:
        count += 1

print(count)
```

## **Mixed Operators**

```python
# Build a simple calculator using arithmetic and comparison operators.

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operator (+,-,*,/): ")

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b)
else:
    print("Invalid Operator")
```

```python
# Create a grading system using comparison and logical operators.

marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
# Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.

print("1.Add")
print("2.Subtract")
print("3.Multiply")
print("4.Divide")

choice = int(input("Enter choice: "))

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

if choice == 1:
    print(a + b)
elif choice == 2:
    print(a - b)
elif choice == 3:
    print(a * b)
elif choice == 4:
    if b != 0:
        print(a / b)
    else:
        print("Division by zero not allowed")
else:
    print("Invalid Choice")
```
